```python
# Crea una nueva tabla (df_year) seleccionando únicamente 
# las columnas "Entity" y "Year" de tu tabla original
df_year = df[["Entity", "Year"]]

# Muestra las primeras 5 filas de esta nueva tabla 
# para que puedas ver cómo luce el resultado
df_year.head()
```

```python
# Filtra la tabla original (df) para quedarse únicamente 
# con las filas donde la columna "Year" sea exactamente igual a 2020
# y guarda ese resultado en una nueva tabla llamada df_extraccion_filas
df_extraccion_filas = df[df["Year"] == 2020]

# Muestra las primeras 5 filas de esta nueva tabla filtrada 
# para comprobar que solo contenga información del año 2020
df_extraccion_filas.head()
```

```python
# Filtra la tabla original aplicando DOS condiciones al mismo tiempo usando '&' (AND):
# 1. Que el año sea 2020 (df["Year"] == 2020)
# 2. Que la entidad sea exactamente "Mexico" (df["Entity"] == "Mexico")
# Guarda el resultado en una nueva tabla llamada df_2020_mx
df_2020_mx = df[(df["Year"]==2020) & (df["Entity"]=="Mexico")]

# Muestra las primeras 5 filas de esta nueva tabla 
# para comprobar que solo tiene la información de México en 2020
df_2020_mx.head()
```

```python
#Rango de fechas desde el año que nacimos hasta el 2020
df_2005_2020_mex = df [(df["Year"] >= 2005) & (df["Year"] <= 2020) & (df["Code"] == "MEX")]
df_2005_2020_mex.head()
```

```python
df_ext_EYA= df_2005_2020_mex[["Entity", "Year","Annual change in population"]]
df_ext_EYA
```

```python
# Calcula el promedio (la media) de cada una de las columnas numéricas de la tabla df_ext_EYA
# numeric_only=True le indica a pandas que ignore las columnas de texto y solo calcule 
# el promedio en las columnas que contienen números para evitar errores
df_ext_EYA.mean(numeric_only=True)
```

```python
# Crea una nueva tabla (df_new) extrayendo únicamente la columna "Entity" 
# de la tabla original (los doble corchetes [[]] aseguran que el resultado se mantenga como tabla)
df_new = df[["Entity"]]

# Muestra el contenido de esta nueva tabla en pantalla
df_new
```

```python
# .loc se usa para seleccionar filas y columnas por su nombre o etiqueta
# El primer ":" antes de la coma significa que queremos TODAS las filas de la tabla
# La lista ["Entity", "Year"] indica que solo queremos conservar esas dos columnas específicas
df.loc[:, ["Entity", "Year"]]
```

```python
# .iloc se usa para seleccionar filas y columnas basándose en su posición numérica (índices)
# El primer ":" antes de la coma significa que queremos TODAS las filas de la tabla
# "0:2" después de la coma indica que queremos las columnas desde la posición 0 hasta la 1 
# (el 2 está excluido, por lo que selecciona exactamente las dos primeras columnas)
df.iloc[:, 0:2]
```

```python
df.loc[30:35]
```

EJERCICIOS loc y iloc

```python
#1 loc- filas: filas cuyo indice sea 10000
df.loc[10000]
```

```python
# 4. loc - filtro . selecciona todos los registros de mexico al año 2020
df.loc[(df["Year"] == 2020) & (df["Entity"] == "Mexico")]
```

```python
#5. iloc - filas: selecciona primera 5 filas del df
df.iloc[0:5]
```

```python
#6. iloc - columnas: secciona todas las filas de las columnas que ocupan las posiciones 1,2 y 3
df.iloc[:, 1:4]
```

```python
#7. iloc - filas y columnas: selecciona las filas de la posicion 0 o 4 y columnas de la posicion 0 a 3
df.iloc[0:5, 0:4]
```

Actividad 3 — Estadística descriptiva

```python
#año 2005-2023 - pais que empiece M -> sera tu df
df_filtrado = df [(df["Year"] >= 2005) & (df["Year"] <= 2023) & (df["Entity"]=="Morocco")]
df_filtrado.head
```

```python
print("Media:", df_filtrado.mean(numeric_only=True) )
```

```python
print("Mediana:", df_filtrado.median(numeric_only=True) )
```

```python
print("Varianza:", df_filtrado.var(numeric_only=True) )
```

```python
print("Desviación estándar:", df_filtrado.std(numeric_only=True) )
```

```python
print("Mínimo:", df_filtrado.min(numeric_only=True))
```

```python
print("Máximo:", df_filtrado.max(numeric_only=True) )
```
