## 5. Preparar los nombres de las columnas

Al explorar el dataset notamos dos cosas:

1. Los nombres de las columnas son largos y difíciles de escribir.
2. Los **datos históricos** (`..._variant_estimates`) y las **proyecciones de la ONU** (`..._variant_medium__projected`) vienen en **columnas separadas**. Cuando una fila es histórica, la columna de proyección está vacía (`NaN`), y viceversa.

Vamos a renombrar las columnas y a crear dos columnas unificadas, `births` y `deaths`, que contengan el dato histórico cuando existe y la proyección de OWID cuando no.

````python
df = df.rename(columns={
    "entity": "Entity",
    "code": "Code",
    "year": "Year",
    "births__sex_all__age_all__variant_estimates": "births_estimates",
    "births__sex_all__age_all__variant_medium__projected": "births_projected",
    "deaths__sex_all__age_all__variant_estimates": "deaths_estimates",
    "deaths__sex_all__age_all__variant_medium__projected": "deaths_projected",
})


df.head()
````

````python
df_inicial = df.copy()
df_inicial.head(3)
````

````python
# Columna unificada: dato histórico si existe; si no, la proyección de OWID
df["births"] = df["births_estimates"].fillna(df["births_projected"])
df["deaths"] = df["deaths_estimates"].fillna(df["deaths_projected"])
df.head()
````

Para nuestro primer modelo utilizaremos:

* **X:** año → la **variable predictora**

Obtenemos los valores como arreglos de NumPy:

### Observación

Scikit-learn espera normalmente que las variables predictoras tengan la forma:

```text
(número de observaciones, número de variables)
```
Es decir, que scikit-learn interpreta X como una tabla de datos.

Por eso `X` tiene **dos dimensiones**, mientras que `y` tiene una sola.

En nuestro ejemplo tenemos:

74 observaciones (registros).

1 variable predictora: año (estas son las caracteristicas o atributos usadas para la predicción).

# PARTE 7. Modelo 1 – Regresión lineal

## 13. Crear el modelo

````python
modelo_lineal = LinearRegression()
````

## 14. Entrenar el modelo

Utilizamos los datos históricos para entrenar el modelo.

````python
modelo_lineal.fit(X, y_nacimientos)
````

Una regresión lineal aprende una recta `y = a·x + b`. Veamos qué valores aprendió:

````python
print("Pendiente (a):", modelo_lineal.coef_[0])
print("Intercepto (b):", modelo_lineal.intercept_)
````

## 15. Realizar predicciones

Utilizamos el modelo para estimar los nacimientos correspondientes a los **años históricos**.

````python
predicciones_lineal = modelo_lineal.predict(X)

predicciones_lineal[:10]
````

## 16. Realizar proyecciones hasta 2100

Creamos los años para los que queremos generar **nuestras propias proyecciones**.

````python
anios_futuros = np.arange(2024, 2101).reshape(-1, 1)

print("Forma:", anios_futuros.shape)
anios_futuros[:10]
````

> **Detalle técnico:** entrenamos el modelo con `X`, que es un `DataFrame` con una columna llamada `Year`. Scikit-learn recuerda ese nombre, así que para predecir también debemos entregarle un `DataFrame` con una columna `Year`. Si le pasáramos un arreglo de NumPy "a secas" mostraría una advertencia (*UserWarning: X does not have valid feature names*). Por eso convertimos los años futuros a `DataFrame`.

````python
X_futuro = pd.DataFrame(anios_futuros, columns=["Year"])

X_futuro.head()
````

````python
predicciones_lineal_futuras = modelo_lineal.predict(X_futuro)
````

````python
print(predicciones_lineal_futuras[:10])
````

# PARTE 8. Visualizar la regresión lineal

## 17. Comparar datos históricos y predicciones

````python
plt.figure(figsize=(12, 6))

plt.scatter(
    df_historico["Year"],
    df_historico["births"],
    label="Datos históricos",
    alpha=0.6
)

# Ajuste del modelo sobre los años históricos
plt.plot(
    df_historico["Year"],
    predicciones_lineal,
    color="tab:orange",
    label="Regresión lineal (ajuste histórico)"
)

# Proyección hacia el futuro
plt.plot(
    anios_futuros.flatten(),
    predicciones_lineal_futuras,
    color="tab:orange",
    linestyle="--",
    label="Regresión lineal (proyección)"
)

plt.title("Proyección de nacimientos – Regresión lineal")
plt.xlabel("Año")
plt.ylabel("Nacimientos")

plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

### Observación

Analiza cómo continúa la tendencia histórica cuando utilizamos una **función lineal**. ¿Qué pasa con los valores hacia el año 2100?
Hacia el año 2100, la función lineal proyecta un crecimiento constante y ascendente que supera los 3 millones de nacimientos. Sin embargo, esto no coincide con la realidad histórica, ya que los datos reales muestran un descenso; esto ocurre porque el modelo lineal es rígido, asume una sola dirección y no puede adaptarse a los cambios de tendencia o curvas del comportamiento demográfico. el modelo no entiende de años ni de historia humana, solo ve números. Para él, como al inicio subió tanto, la línea debe inclinarse hacia arriba. Cuando llega la época moderna y los datos reales empiezan a bajar, la línea recta ya no puede doblarse ni curvarse hacia abajo porque su propia estructura matemática se lo prohíbe; está obligada a seguir la misma inclinación recta con la que empezó.

# PARTE 9. Modelo 2 – Regresión polinomial

## 18. Crear variables polinomiales

La regresión lineal solamente puede representar una relación **lineal** entre las variables. Ahora construiremos variables polinomiales para permitir **curvas**. Utilizaremos un polinomio de **grado 2**:

$$y = a \cdot x^2 + b \cdot x + c$$

````python
poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)
````

Transformamos `X`:

````python
X_poly = poly.fit_transform(X)

X_poly[:5]
````

Observa que ahora cada año tiene **dos columnas**: el año (`x`) y el año al cuadrado (`x²`).

## 19. Crear y entrenar el modelo polinomial

Seguimos utilizando `LinearRegression`; lo único que cambió son las **variables de entrada**.

````python
modelo_polinomial = LinearRegression()

modelo_polinomial.fit(
    X_poly,
    y_nacimientos
)
````

## 20. Generar predicciones históricas

````python
predicciones_polinomiales = modelo_polinomial.predict(
    X_poly
)
````

## 21. Generar proyecciones hasta 2100

Transformamos los años futuros utilizando **el mismo objeto `poly`** con el que transformamos los datos de entrenamiento.

````python
anios_futuros_poly = poly.transform(
    X_futuro
)

predicciones_polinomiales_futuras = modelo_polinomial.predict(
    anios_futuros_poly
)
````

# PARTE 10. Visualizar la regresión polinomial

## 22. Graficar las predicciones

````python
plt.figure(figsize=(12, 6))

plt.scatter(
    df_historico["Year"],
    df_historico["births"],
    label="Datos históricos",
    alpha=0.6
)

plt.plot(
    df_historico["Year"],
    predicciones_polinomiales,
    color="tab:green",
    label="Regresión polinomial (ajuste histórico)"
)

plt.plot(
    anios_futuros.flatten(),
    predicciones_polinomiales_futuras,
    color="tab:green",
    linestyle="--",
    label="Regresión polinomial (proyección)"
)

plt.title("Proyección de nacimientos – Regresión polinomial")
plt.xlabel("Año")
plt.ylabel("Nacimientos")

plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

### Observación

Compara el comportamiento de esta gráfica con el modelo de regresión lineal. ¿Se ajusta mejor a los datos históricos? Sí. A diferencia de la la regresión lineal anterior, la curva de la regresión polinomial sí logra doblarse y seguir de cerca la forma real de los datos historicos en el pasado 

¿Qué ocurre con la proyección hacia el año 2100? La proyección regresión polinomial se desploma de forma pronunciada conforme se acerca al año 2100, cayendo hasta números negativos.

¿Los valores que predice tienen sentido para un número de nacimientos? No, no tienen sentido. No puede haber un número negativo de nacimientos (no pueden nacer "-4 millones" de personas). Esto demuestra que aunque el modelo polinomial captura bien el pasado, falla  al proyectar a largo plazo porque la curva se acelera sin control hacia abajo.

# PARTE 11. Modelo 3 – Random Forest

## 23. Crear el modelo

Un **Random Forest** está formado por muchos árboles de decisión. Utilizaremos `RandomForestRegressor`.

````python
modelo_rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
````

## 24. Entrenar el modelo

````python
modelo_rf.fit(
    X,
    y_nacimientos
)
````

## 25. Generar predicciones históricas

````python
predicciones_rf = modelo_rf.predict(X)
````

## 26. Generar proyecciones futuras

````python
predicciones_rf_futuras = modelo_rf.predict(
    X_futuro
)
````

# PARTE 12. Visualizar Random Forest

## 27. Graficar las predicciones

````python
plt.figure(figsize=(12, 6))

plt.scatter(
    df_historico["Year"],
    df_historico["births"],
    label="Datos históricos",
    alpha=0.6
)

plt.plot(
    df_historico["Year"],
    predicciones_rf,
    color="tab:red",
    label="Random Forest (ajuste histórico)"
)

plt.plot(
    anios_futuros.flatten(),
    predicciones_rf_futuras,
    color="tab:red",
    linestyle="--",
    label="Random Forest (proyección)"
)

plt.title("Proyección de nacimientos – Random Forest")
plt.xlabel("Año")
plt.ylabel("Nacimientos")

plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

### Observación

Fíjate en dos cosas: (1) qué tan bien sigue el modelo los datos históricos
Se ajusta de manera perfecta, sigue la linea de los datos historicos
 (2) qué sucede con la línea después de 2023. ¿Continúa la tendencia? No, continua de manera horizontal
 
 ¿Por qué crees que sucede esto? Investiga.
 Esto ocurre por la propia naturaleza de cómo funciona un Random Forest (que está compuesto por múltiples árboles de decisión):

-Incapacidad de extrapolar: Los árboles de decisión hacen predicciones dividiendo los datos y tomando promedios de los valores que ya conocen en sus nodos finales.
-Límite del entrenamiento: Como el modelo solo fue entrenado con años del pasado (hasta 2023), no sabe "inventar" ni proyectar valores que vayan más allá de ese rango máximo.
-Al pedirle que prediga años futuros (como el 2100), el algoritmo cae en el nodo final y simplemente repite o mantiene constante el valor promedio del último grupo de datos conocidos, dando como resultado esa línea continua.

````python
# extraer china
pais = "China"
df_china = df[df["Entity"] == pais].copy()

df_china.head()
````

````python
#SEPARAR HISTÓRICO/PROYECTADO
df_historico = df_china[df_china["Year"] <= 2023]
df_proyectado = df_china[df_china["Year"] > 2023]

print(f"Histórico de {pais} ->", df_historico.shape, "| Años:", df_historico["Year"].min(), "a", df_historico["Year"].max())
print(f"Proyectado de {pais} ->", df_proyectado.shape, "| Años:", df_proyectado["Year"].min(), "a", df_proyectado["Year"].max())
````

````python
#PREPARAR VARIABLES PARA MACHINE LEARNING
X = df_historico[["Year"]]
y_nacimientos = df_historico["births"]
y_defunciones = df_historico["deaths"]

anios_futuros = np.arange(2024, 2101).reshape(-1, 1)
X_futuro = pd.DataFrame(anios_futuros, columns=["Year"])
````

````python
#MODELO 1: REGRESIÓN LINEAL
modelo_lineal = LinearRegression()
modelo_lineal.fit(X, y_nacimientos)
predicciones_lineal_futuras = modelo_lineal.predict(X_futuro)

modelo_lineal_deaths = LinearRegression()
modelo_lineal_deaths.fit(X, y_defunciones)
predicciones_lineal_deaths = modelo_lineal_deaths.predict(X_futuro)
````

````python
#MODELO 2: REGRESIÓN POLINOMIAL 
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)
anios_futuros_poly = poly.transform(X_futuro)

modelo_polinomial = LinearRegression()
modelo_polinomial.fit(X_poly, y_nacimientos)
predicciones_polinomiales_futuras = modelo_polinomial.predict(anios_futuros_poly)

modelo_polinomial_deaths = LinearRegression()
modelo_polinomial_deaths.fit(X_poly, y_defunciones)
predicciones_polinomiales_deaths = modelo_polinomial_deaths.predict(anios_futuros_poly)
````

````python
#MODELO 3: RANDOM FOREST REGRESSOR
modelo_rf = RandomForestRegressor(n_estimators=100, random_state=42)
modelo_rf.fit(X, y_nacimientos)
predicciones_rf_futuras = modelo_rf.predict(X_futuro)

modelo_rf_deaths = RandomForestRegressor(n_estimators=100, random_state=42)
modelo_rf_deaths.fit(X, y_defunciones)
predicciones_rf_deaths = modelo_rf_deaths.predict(X_futuro)
````

````python
#GRÁFICA COMPARATIVA DE LOS 3 MODELOS (NACIMIENTOS - CHINA)
plt.figure(figsize=(14, 7))
plt.scatter(df_historico["Year"], df_historico["births"], label="Datos históricos", alpha=0.5, color="black")
plt.plot(anios_futuros.flatten(), predicciones_lineal_futuras, color="tab:orange", label="Regresión lineal")
plt.plot(anios_futuros.flatten(), predicciones_polinomiales_futuras, color="tab:green", label="Regresión polinomial")
plt.plot(anios_futuros.flatten(), pred_rf_futuras if 'pred_rf_futuras' in locals() else predicciones_rf_futuras, color="tab:red", label="Random Forest")
plt.title("Comparación de modelos para proyectar nacimientos en China")
plt.xlabel("Año")
plt.ylabel("Nacimientos")
plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````

````python
#GRÁFICA COMPARATIVA DE LOS 3 MODELOS (DEFUNCIONES - CHINA)
plt.figure(figsize=(14, 7))
plt.scatter(df_historico["Year"], df_historico["deaths"], label="Datos históricos", alpha=0.5, color="black")
plt.plot(anios_futuros.flatten(), predicciones_lineal_deaths, color="tab:orange", label="Regresión lineal")
plt.plot(anios_futuros.flatten(), predicciones_polinomiales_deaths, color="tab:green", label="Regresión polinomial")
plt.plot(anios_futuros.flatten(), predicciones_rf_deaths, color="tab:red", label="Random Forest")
plt.title("Comparación de modelos para proyectar defunciones en China")
plt.xlabel("Año")
plt.ylabel("Defunciones")
plt.ticklabel_format(style="plain", axis="y", useOffset=False)
plt.legend()
plt.grid()
plt.show()
````
